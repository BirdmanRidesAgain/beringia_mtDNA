### Annotating different levels of completeness

Because NOVOPlasty sometimes fails to fully assemble a mitogenome (or produces fragmented assemblies), four different types of input went into the annotation pipeline.
Levels 1 and 2, corresponding to `success` typically resulted in good-quality annotations.
Level 3 also had the annotator run, but often did not result in usable data.
Level 4 had no outputs which the annotator could actually be run - these did not produce any annotation data.

| Tier | Name | Description | Number |
| - | - | - | - |
| 1 | `Circularized_assembly_*.fasta`| Unambiguously resolved mitogenome. Good-quality annotations possible. | 147 |
| 2 | `Option_1_*.fasta` | Plausible assembly, but more than one ordering present. Good-quality annotations possible. | 19 `Option_1`, 3 `Option_1` + `Option_2` |
| 3 | `Contigs_*.fasta` | One or more contigs present. May result in partial annotations. | 150 |
| 4 | `contigs_tmp_*.txt` | Individual, small contigs. Most have file sizes of 0. Cannot be annotated. | 93 |

## Code documentation

We run `mitoz` through the conda environment `mitoz-x64`.
On an OSX machine, you install it with the conda environment in [mitoz-x64.yml](resources/mitoz-x64.yml).

We then ran the program using [annotate_mitogenome.py](scripts/annotate_mitogenome.py) to call `mitoz`, and [batch_annotate_assemblies.py](scripts/batch_annotate_assemblies.py) to iterate across all files in the directory.

```bash
# sample code for a single run
conda activate mitoz-x64 
python scripts/annotate_mitogenome.py -i $FA -s sample1 -v
```

Outputs are under `mitogenomes_output/<sample>/annotation/<suffix>/mitoz/`.
A log file can be found at [batch_annotate_assemblies.log](../mitogenomes_output/_batch_logs/batch_annotate_assemblies.log).
